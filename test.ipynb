This notebook runs the tests in src/tests folder using pytest.

Test runner: set `TESTS` to an empty list to run everything, or list test files or single tests (`file.py::test_name`). Set `LIST_ONLY = True` to see every available test name.

```python
# ---- Test runner: edit the settings below, then run this cell ----
# TESTS: leave empty to run ALL tests, or list files / single tests, e.g.
#   ["src/tests/test_models.py"]
#   ["src/tests/test_models.py::test_get_params_returns_dict", "src/tests/test_split.py"]
TESTS = []
KEYWORD = ""          # optional -k filter, e.g. "shape" or "not baseline"
VERBOSE = True        # -v
STOP_ON_FIRST = False # -x
LIST_ONLY = False     # True = just list available tests, don't run them

import subprocess, sys

cmd = [sys.executable, "-m", "pytest", *(TESTS or ["src/tests"]), "--color=yes"]
if LIST_ONLY:
    cmd += ["--collect-only", "-q"]
else:
    if VERBOSE:
        cmd.append("-v")
    if STOP_ON_FIRST:
        cmd.append("-x")
    if KEYWORD.strip():
        cmd += ["-k", KEYWORD.strip()]

print("$ python " + " ".join(cmd[1:]) + "\n", flush=True)
proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                        stdin=subprocess.DEVNULL, text=True, encoding="utf-8", errors="replace")
for line in proc.stdout:          # stream output live as tests run
    print(line, end="", flush=True)
proc.wait()
print(f"\nExit code: {proc.returncode}  (0 = all passed)")
```
